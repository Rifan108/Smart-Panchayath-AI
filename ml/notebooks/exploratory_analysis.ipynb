# Exploratory Data Analysis: Smart Panchayat Socioeconomic Dynamics

This notebook analyzes the 60,000-record synthetic Panchayat dataset calibrated on official benchmarks (PAI, NITI Aayog MPI, NFHS-5, Census 2011) across **Kerala, Karnataka, Tamil Nadu, Andhra Pradesh, and Maharashtra**.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_parquet('../datasets/synthetic/synthetic_panchayats_all_5_states.parquet')
print(f'Dataset Shape: {df.shape}')
df.head()
```

## 1. Cross-State Comparative Distribution of Core Indicators


```python
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
sns.boxplot(data=df, x='state', y='poverty_rate', ax=axes[0, 0], palette='Set2')
axes[0, 0].set_title('Multidimensional Poverty Rate by State')
axes[0, 0].tick_params(axis='x', rotation=20)

sns.boxplot(data=df, x='state', y='drinking_water_coverage', ax=axes[0, 1], palette='Set2')
axes[0, 1].set_title('Drinking Water Tap Coverage (JJM) by State')
axes[0, 1].tick_params(axis='x', rotation=20)

sns.boxplot(data=df, x='state', y='literacy_rate', ax=axes[1, 0], palette='Set2')
axes[1, 0].set_title('Adult Literacy Rate by State')
axes[1, 0].tick_params(axis='x', rotation=20)

sns.boxplot(data=df, x='state', y='annual_panchayat_budget_lakhs', ax=axes[1, 1], palette='Set2')
axes[1, 1].set_title('Annual Panchayat Budget (₹ Lakhs) by State')
axes[1, 1].tick_params(axis='x', rotation=20)

plt.tight_layout()
plt.show()
```

## 2. Priority Sector Distribution Across States


```python
pd.crosstab(df['state'], df['priority_sector'], normalize='index').round(3) * 100
```

## 3. Correlation Structure Among Domain Scores


```python
score_cols = [c for c in df.columns if 'score' in c]
plt.figure(figsize=(10, 8))
sns.heatmap(df[score_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix: Domain Scores')
plt.show()
```
